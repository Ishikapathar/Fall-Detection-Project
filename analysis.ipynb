```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('fall_detection_dataset.csv')
df.columns = df.columns.str.strip().str.lower()

# Room Temperature with color
plt.figure()
df['room_temp'].hist(color='skyblue')
plt.title('Distribution of Room Temperature')
plt.xlabel('Room Temperature')
plt.ylabel('Frequency')
plt.show()

# Room Light with color
plt.figure()
df['room_light'].hist(color='orange')
plt.title('Distribution of Room Light')
plt.xlabel('Room Light')
plt.ylabel('Frequency')
plt.show()

# Time of Day with different colors
plt.figure()
df['time_of_day'].value_counts().plot(kind='bar', color=['red', 'green', 'blue', 'purple'])
plt.title('Distribution of Time of Day')
plt.xlabel('Time of Day')
plt.ylabel('Count')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('fall_detection_dataset.csv')
df.columns = df.columns.str.strip().str.lower()

# Room Temp vs Fall Event
plt.figure()
for val, color in zip([0,1], ['blue','red']):
    subset = df[df['fall_event']==val]
    plt.hist(subset['room_temp'], alpha=0.5, label=f'fall_event={val}')
plt.title('Room Temp vs Fall Event')
plt.xlabel('Room Temp')
plt.ylabel('Frequency')
plt.legend()
plt.show()

# Room Light vs Fall Event
plt.figure()
for val, color in zip([0,1], ['green','orange']):
    subset = df[df['fall_event']==val]
    plt.hist(subset['room_light'], alpha=0.5, label=f'fall_event={val}')
plt.title('Room Light vs Fall Event')
plt.xlabel('Room Light')
plt.ylabel('Frequency')
plt.legend()
plt.show()

# Time of Day vs Fall Event
plt.figure()
pd.crosstab(df['time_of_day'], df['fall_event']).plot(kind='bar')
plt.title('Time of Day vs Fall Event')
plt.xlabel('Time of Day')
plt.ylabel('Count')
plt.show()
```
